**Analysis**:This is for my job simulation. I have been given a task to build models that are able to determine the fair price of coffee option contract

```python
#import numpy
import numpy as np
#we now we look at the various data required for the models
S_T = 1.20
r= 0.02
d= 0.01
T=0.5
X=1.25
vol= 0.25
```

**COST OF CARRY MODEL**

```python
#USING the above data. I will create the cost of carry model
#we will calculate the future prices based on the above variables
future_price = S_T * np.exp((r+d)*T)

#we will print it out
print(f"Coffee fair price contract : ${future_price:.4f}")
```

**Black-Scholes Model**

```python
from scipy.stats import norm
#we iwill build our black-scholes model:
#finding c_1 and c_2 to simplify the calculations
c_1 = (np.log(S_T/X) + (r+T*vol**2)*T)/(vol*np.sqrt(T))
c_2= c_1-vol*np.sqrt(T)
call_option_price = S_T *norm.cdf(c_1) - X * np.exp(-r*T) * norm.cdf(c_2)

#now we print
print(f"Coffee fair price contract : ${call_option_price:.4f}")
#
```

**Monte Carlo Simulation**

```python
#add few new variables:
simulations_count = 10000
step_count= 252

#incremenation
increment = T/step_count

#we areexepected to have reproductivity
np.random.seed(42)
paths_price= np.zeros((step_count, simulations_count))
paths_price[0]= S_T
 #begin the loop
for u in range(1, step_count):
  o = np.random.standard_normal(simulations_count)
  paths_price[u]= paths_price[u-1]*np.exp((r-0.5-vol**2)*increment + vol*np.sqrt(increment)*o)

  #calculate avrerage simulated price when it hits maturity

simulated_price_average= np.mean(paths_price[-1])
print(f"Average simulated price of the coffee : ${simulated_price_average:.4f}")
#
```
